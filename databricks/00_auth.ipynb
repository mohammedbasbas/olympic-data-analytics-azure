# 🔐 00 – Authentification Databricks → ADLS Gen2

Configure l'accès OAuth 2.0 (Service Principal) au Data Lake.

- Le **client secret** est lu depuis **Azure Key Vault** via un *Key Vault-backed secret scope* (`secretScope` / `secretKey`) : aucun secret en clair.
- Remplacer `<CLIENT_ID>` (Application ID de l'App Registration) et `<TENANT_ID>` (Directory ID Entra ID).
- Ce notebook est appelé par les autres via `%run ./00_auth`.

```python
service_credential = dbutils.secrets.get(scope="secretScope",key="secretKey")

spark.conf.set("fs.azure.account.auth.type.mohammedstorageacc2.dfs.core.windows.net", "OAuth")
spark.conf.set("fs.azure.account.oauth.provider.type.mohammedstorageacc2.dfs.core.windows.net", "org.apache.hadoop.fs.azurebfs.oauth2.ClientCredsTokenProvider")
spark.conf.set("fs.azure.account.oauth2.client.id.mohammedstorageacc2.dfs.core.windows.net", "<CLIENT_ID>")
spark.conf.set("fs.azure.account.oauth2.client.secret.mohammedstorageacc2.dfs.core.windows.net", service_credential)
spark.conf.set("fs.azure.account.oauth2.client.endpoint.mohammedstorageacc2.dfs.core.windows.net", "https://login.microsoftonline.com/<TENANT_ID>/oauth2/token")
```
